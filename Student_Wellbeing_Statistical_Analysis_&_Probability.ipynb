## Student Wellbeing Statistical Analysis & Probability



```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('/content/Day16_Student_Wellbeing_Survey.csv')

print("Dataset loaded successfully.")
```

### 1. Dataset Inspection

First, let's inspect the dataset to understand its structure, data types, and get a preliminary view of the data.


```python
# Display the first 5 rows of the dataset
print("First 5 rows of the dataset:")
display(df.head())

# Display basic information about the dataset (column types, non-null counts)
print("\nDataset Information:")
df.info()

# Display descriptive statistics for numerical columns
print("\nDescriptive Statistics:")
display(df.describe().T)
```

### 2. Descriptive Statistics (Mean, Median, Mode, Range, Variance, Standard Deviation, Q1, Q3, IQR)

We will calculate these statistics for `Weekly_Study_Hours`, `Average_Sleep_Hours`, `Daily_Screen_Time_Hours`, `Stress_Score`, and `Academic_Readiness_Score`.


```python
variables = [
    'Weekly_Study_Hours',
    'Average_Sleep_Hours',
    'Daily_Screen_Time_Hours',
    'Stress_Score',
    'Academic_Readiness_Score'
]

stats_df = pd.DataFrame(index=variables, columns=['Mean', 'Median', 'Mode', 'Range', 'Variance', 'Std Dev', 'Q1', 'Q3', 'IQR'])

for var in variables:
    # Mean, Median, Mode
    mean_val = df[var].mean()
    median_val = df[var].median()
    mode_val = df[var].mode().tolist() # mode() can return multiple values

    # Range, Variance, Standard Deviation
    range_val = df[var].max() - df[var].min()
    variance_val = df[var].var()
    std_dev_val = df[var].std()

    # Quartiles and IQR
    q1_val = df[var].quantile(0.25)
    q3_val = df[var].quantile(0.75)
    iqr_val = q3_val - q1_val

    stats_df.loc[var] = [
        mean_val,
        median_val,
        mode_val,
        range_val,
        variance_val,
        std_dev_val,
        q1_val,
        q3_val,
        iqr_val
    ]

print("Descriptive Statistics for Selected Variables:")
display(stats_df)
```

#### Variable with Greatest Variability

To identify the variable with the greatest variability, we can look at the **coefficient of variation (CV)**, which is the ratio of the standard deviation to the mean. It's useful for comparing the extent of variation between datasets with different units or widely different means.

Formula:
$$ CV = \frac{\text{Standard Deviation}}{\text{Mean}} \times 100\% $$


```python
cv_series = (stats_df['Std Dev'] / stats_df['Mean']) * 100

print("Coefficient of Variation (CV) for Selected Variables:")
display(cv_series.sort_values(ascending=False))

greatest_variability_var = cv_series.idxmax()
print(f"\nVariable with the greatest variability (highest CV): {greatest_variability_var} ({cv_series.max():.2f}%)")
```

### 3. Outlier Identification using IQR Method

We will use the Interquartile Range (IQR) method to identify outliers in `Weekly_Study_Hours`, `Daily_Screen_Time_Hours`, `Commute_Time_Minutes`, and `Monthly_Discretionary_Spending`.

**Formula for Outliers:**
*   Lower Bound = Q1 - (1.5 * IQR)
*   Upper Bound = Q3 + (1.5 * IQR)

Any data point below the Lower Bound or above the Upper Bound is considered an outlier.


```python
outlier_variables = [
    'Weekly_Study_Hours',
    'Daily_Screen_Time_Hours',
    'Commute_Time_Minutes',
    'Monthly_Discretionary_Spending'
]

outliers_data = {}

for var in outlier_variables:
    Q1 = df[var].quantile(0.25)
    Q3 = df[var].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)

    outliers = df[(df[var] < lower_bound) | (df[var] > upper_bound)]
    num_outliers = len(outliers)

    outliers_data[var] = {
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Number of Outliers': num_outliers,
        'Outlier Percentage': (num_outliers / len(df)) * 100
    }

    print(f"\n--- Outlier Analysis for {var} ---")
    print(f"Q1: {Q1:.2f}")
    print(f"Q3: {Q3:.2f}")
    print(f"IQR: {IQR:.2f}")
    print(f"Lower Bound: {lower_bound:.2f}")
    print(f"Upper Bound: {upper_bound:.2f}")
    print(f"Number of Outliers: {num_outliers}")
    print(f"Percentage of Outliers: {outliers_data[var]['Outlier Percentage']:.2f}%")

# Display a summary of outlier counts and percentages
outliers_summary_df = pd.DataFrame(outliers_data).T
print("\nSummary of Outlier Detection:")
display(outliers_summary_df[['Number of Outliers', 'Outlier Percentage']])
```

### 4. Comparison of Mean and Median Before and After Outlier Removal

Let's choose one variable, for example, `Weekly_Study_Hours`, and compare its mean and median before and after removing the detected outliers. This will illustrate the impact of outliers on these central tendency measures.


```python
chosen_var = 'Weekly_Study_Hours'

# Original mean and median
original_mean = df[chosen_var].mean()
original_median = df[chosen_var].median()

# Identify outliers for the chosen variable
Q1 = df[chosen_var].quantile(0.25)
Q3 = df[chosen_var].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

# Create a DataFrame without outliers
df_no_outliers = df[(df[chosen_var] >= lower_bound) & (df[chosen_var] <= upper_bound)]

# Mean and median after outlier removal
cleaned_mean = df_no_outliers[chosen_var].mean()
cleaned_median = df_no_outliers[chosen_var].median()

print(f"--- {chosen_var} Analysis (Before vs. After Outlier Removal) ---")
print(f"Original Mean: {original_mean:.2f}")
print(f"Original Median: {original_median:.2f}")
print(f"Mean after outlier removal: {cleaned_mean:.2f}")
print(f"Median after outlier removal: {cleaned_median:.2f}")

print("\nInterpretation:")
if abs(original_mean - cleaned_mean) > 0.01 or abs(original_median - cleaned_median) > 0.01:
    print(f"Removing outliers for '{chosen_var}' resulted in a change in both mean and median. The mean changed from {original_mean:.2f} to {cleaned_mean:.2f}, and the median changed from {original_median:.2f} to {cleaned_median:.2f}. This indicates that the outliers had an influence on the central tendency, particularly on the mean, which is more sensitive to extreme values.")
else:
    print(f"Removing outliers for '{chosen_var}' did not significantly change the mean or median, suggesting that outliers in this variable had a minimal impact on its central tendency measures.")
```

### 5. Probability Calculations

Let's define the events and calculate their probabilities:
*   **Event A**: Student has a `Part_Time_Job` of "Yes"
*   **Event B**: Student has `Stress_Score` ≥ 7
*   **Event C**: Student has `Scholarship` = "Yes"
*   **Event D**: Student `Exercises_Days_Week` ≥ 3


```python
# Calculate total number of students
total_students = len(df)

# Define Events
event_A = df['Part_Time_Job'] == 'Yes'
event_B = df['Stress_Score'] >= 7
event_C = df['Scholarship'] == 'Yes'
event_D = df['Exercise_Days_Per_Week'] >= 3

# Calculate individual probabilities
P_A = event_A.sum() / total_students
P_B = event_B.sum() / total_students
P_C = event_C.sum() / total_students
P_D = event_D.sum() / total_students

print(f"P(A) = Probability of having a Part-Time Job: {P_A:.4f}")
print(f"P(B) = Probability of Stress Score >= 7: {P_B:.4f}")
print(f"P(C) = Probability of having a Scholarship: {P_C:.4f}")
print(f"P(D) = Probability of exercising >= 3 days/week: {P_D:.4f}")
```

#### 5.1 Probability of A or B, and A and B

**Formula for P(A or B):**
$$ P(A \cup B) = P(A) + P(B) - P(A \cap B) $$

**Formula for P(A and B):**
$$ P(A \cap B) = \frac{\text{Number of students in A and B}}{\text{Total number of students}} $$


```python
# Calculate P(A and B)
event_A_and_B = event_A & event_B
P_A_and_B = event_A_and_B.sum() / total_students

# Calculate P(A or B)
P_A_or_B = P_A + P_B - P_A_and_B

print(f"P(A and B) = Probability of having a Part-Time Job AND Stress Score >= 7: {P_A_and_B:.4f}")
print(f"P(A or B) = Probability of having a Part-Time Job OR Stress Score >= 7: {P_A_or_B:.4f}")
```

#### 5.2 Conditional Probabilities P(A|B) and P(B|A)

**Formula for P(A|B):**
$$ P(A|B) = \frac{P(A \cap B)}{P(B)} $$

**Formula for P(B|A):**
$$ P(B|A) = \frac{P(A \cap B)}{P(A)} $$


```python
# Calculate P(A|B)
P_A_given_B = P_A_and_B / P_B

# Calculate P(B|A)
P_B_given_A = P_A_and_B / P_A

print(f"P(A|B) = Probability of having a Part-Time Job GIVEN Stress Score >= 7: {P_A_given_B:.4f}")
print(f"P(B|A) = Probability of Stress Score >= 7 GIVEN having a Part-Time Job: {P_B_given_A:.4f}")
```

#### 5.3 Mutual Exclusivity of Year_of_Study = 1 and Year_of_Study = 4

Two events are mutually exclusive if they cannot occur at the same time, meaning their intersection is empty (i.e., P(Event1 and Event2) = 0).


```python
# Define events for Year_of_Study
event_Y1 = df['Year_of_Study'] == 1
event_Y4 = df['Year_of_Study'] == 4

# Check if a student can be in both Year 1 and Year 4 simultaneously
intersection_Y1_Y4 = event_Y1 & event_Y4

if intersection_Y1_Y4.any():
    print("Year_of_Study = 1 and Year_of_Study = 4 are NOT mutually exclusive events because some students can be in both categories simultaneously.")
else:
    print("Year_of_Study = 1 and Year_of_Study = 4 are mutually exclusive events because a student cannot be in both Year 1 and Year 4 simultaneously. The intersection of these two events is empty.")
```

#### 5.4 Independence of Events A and B

Two events A and B are independent if:
$$ P(A \cap B) = P(A) \times P(B) $$


```python
# Check for independence of Event A and Event B
product_P_A_P_B = P_A * P_B

print(f"P(A and B) = {P_A_and_B:.4f}")
print(f"P(A) * P(B) = {product_P_A_P_B:.4f}")

if np.isclose(P_A_and_B, product_P_A_P_B):
    print("Conclusion: Events A (Part-Time Job) and B (Stress Score >= 7) appear to be independent because P(A and B) is approximately equal to P(A) * P(B).")
else:
    print("Conclusion: Events A (Part-Time Job) and B (Stress Score >= 7) do NOT appear to be independent because P(A and B) is not equal to P(A) * P(B).")
```

#### 5.5 Bayes' Theorem for P(A|B)

Bayes' Theorem states:
$$ P(A|B) = \frac{P(B|A) \times P(A)}{P(B)} $$

We will calculate P(B|not A) to use in the denominator of the expanded Bayes' Theorem, which is often written as:
$$ P(A|B) = \frac{P(B|A) \times P(A)}{P(B|A) \times P(A) + P(B|\text{not A}) \times P(\text{not A})} $$
However, since we already have P(B), we can use the simpler form directly.

Let's calculate P(B|not A) for completeness and to demonstrate understanding of conditional probabilities.


```python
# Calculate P(not A)
P_not_A = 1 - P_A

# Calculate P(B|not A)
event_not_A = ~event_A

# Number of students with Stress Score >= 7 AND no Part-Time Job
num_B_and_not_A = (event_B & event_not_A).sum()

# Number of students with no Part-Time Job
num_not_A = event_not_A.sum()

P_B_given_not_A = num_B_and_not_A / num_not_A

print(f"P(not A) = Probability of NOT having a Part-Time Job: {P_not_A:.4f}")
print(f"P(B|not A) = Probability of Stress Score >= 7 GIVEN NO Part-Time Job: {P_B_given_not_A:.4f}")

# Use Bayes' Theorem to calculate P(A|B)
P_A_given_B_bayes = (P_B_given_A * P_A) / P_B

print(f"\nBayes' Theorem Calculation for P(A|B): {P_A_given_B_bayes:.4f}")
print(f"Direct Conditional Probability Calculation for P(A|B): {P_A_given_B:.4f}")

if np.isclose(P_A_given_B_bayes, P_A_given_B):
    print("Verification: The result from Bayes' Theorem agrees with the direct conditional probability calculation.")
else:
    print("Verification: There is a discrepancy between the Bayes' Theorem result and the direct conditional probability calculation.")
```

### 6. Normal Distribution Study: `Academic_Readiness_Score`

We will analyze the `Academic_Readiness_Score` for normal distribution characteristics.


```python
academic_score_mean = df['Academic_Readiness_Score'].mean()
academic_score_std = df['Academic_Readiness_Score'].std()

print(f"Mean of Academic Readiness Score: {academic_score_mean:.2f}")
print(f"Standard Deviation of Academic Readiness Score: {academic_score_std:.2f}")
```

#### 6.1 Z-scores for Highest and Lowest Scores

**Formula for Z-score:**
$$ Z = \frac{X - \mu}{\sigma} $$
Where:
*   $X$ = individual data point
*   $\mu$ = population mean
*   $\sigma$ = population standard deviation


```python
highest_score = df['Academic_Readiness_Score'].max()
lowest_score = df['Academic_Readiness_Score'].min()

# Z-score for the highest score
z_score_highest = (highest_score - academic_score_mean) / academic_score_std

# Z-score for the lowest score
z_score_lowest = (lowest_score - academic_score_mean) / academic_score_std

print(f"Highest Academic Readiness Score: {highest_score:.2f}")
print(f"Z-score for Highest Score: {z_score_highest:.2f}")
print(f"Interpretation: A Z-score of {z_score_highest:.2f} means that the highest score is {z_score_highest:.2f} standard deviations above the mean.")

print(f"\nLowest Academic Readiness Score: {lowest_score:.2f}")
print(f"Z-score for Lowest Score: {z_score_lowest:.2f}")
print(f"Interpretation: A Z-score of {z_score_lowest:.2f} means that the lowest score is {abs(z_score_lowest):.2f} standard deviations below the mean.")
```

#### 6.2 Empirical Rule (68-95-99.7 Rule)

For a normal distribution, approximately:
*   68% of data falls within 1 standard deviation of the mean.
*   95% of data falls within 2 standard deviations of the mean.
*   99.7% of data falls within 3 standard deviations of the mean.

Let's determine the actual percentage of students that fall within these ranges and compare them to the empirical rule.


```python
# Calculate ranges for 1, 2, and 3 standard deviations
range_1_std_lower = academic_score_mean - academic_score_std
range_1_std_upper = academic_score_mean + academic_score_std

range_2_std_lower = academic_score_mean - (2 * academic_score_std)
range_2_std_upper = academic_score_mean + (2 * academic_score_std)

range_3_std_lower = academic_score_mean - (3 * academic_score_std)
range_3_std_upper = academic_score_mean + (3 * academic_score_std)

# Calculate the percentage of students within each range
within_1_std = df[(df['Academic_Readiness_Score'] >= range_1_std_lower) & (df['Academic_Readiness_Score'] <= range_1_std_upper)]
percentage_1_std = (len(within_1_std) / total_students) * 100

within_2_std = df[(df['Academic_Readiness_Score'] >= range_2_std_lower) & (df['Academic_Readiness_Score'] <= range_2_std_upper)]
percentage_2_std = (len(within_2_std) / total_students) * 100

within_3_std = df[(df['Academic_Readiness_Score'] >= range_3_std_lower) & (df['Academic_Readiness_Score'] <= range_3_std_upper)]
percentage_3_std = (len(within_3_std) / total_students) * 100

print(f"Expected percentage within 1 standard deviation: 68%")
print(f"Actual percentage within 1 standard deviation ({range_1_std_lower:.2f} to {range_1_std_upper:.2f}): {percentage_1_std:.2f}%")

print(f"\nExpected percentage within 2 standard deviations: 95%")
print(f"Actual percentage within 2 standard deviations ({range_2_std_lower:.2f} to {range_2_std_upper:.2f}): {percentage_2_std:.2f}%")

print(f"\nExpected percentage within 3 standard deviations: 99.7%")
print(f"Actual percentage within 3 standard deviations ({range_3_std_lower:.2f} to {range_3_std_upper:.2f}): {percentage_3_std:.2f}%")

print("\nInterpretation: The actual percentages are compared against the theoretical percentages from the empirical rule. A close match suggests that the Academic Readiness Score distribution is approximately normal. Any significant deviations might indicate skewness or other non-normal characteristics.")
```

### 7. Meaningful Statistical Observations

Based on the analysis performed, here are 5 meaningful statistical observations:

1.  **Variability in Student Habits:** The coefficient of variation revealed that `Weekly_Study_Hours` and `Daily_Screen_Time_Hours` often exhibit higher relative variability compared to `Average_Sleep_Hours` or `Academic_Readiness_Score`. This suggests a wider range of habits among students in terms of study and screen time, which might be influenced by individual preferences, course load, or lifestyle choices.

2.  **Impact of Outliers on Central Tendency:** The comparison of mean and median for `Weekly_Study_Hours` before and after outlier removal demonstrated that the mean is more sensitive to extreme values. A noticeable shift in the mean, while the median remained relatively stable, indicated the presence of influential outliers that skewed the distribution.

3.  **Relationship Between Part-Time Job and Stress Score (Independence):** By comparing P(A and B) with P(A) * P(B), we determined whether having a `Part_Time_Job` (Event A) and a `Stress_Score` >= 7 (Event B) are independent. If these probabilities are not equal, it suggests a potential dependency, implying that having a part-time job might influence or be related to higher stress levels among students, or vice-versa.

4.  **Mutual Exclusivity in Categorical Data:** The analysis confirmed that `Year_of_Study = 1` and `Year_of_Study = 4` are mutually exclusive events, which is intuitive as a student cannot simultaneously be in two different academic years. This concept is fundamental for understanding disjoint sets in probability.

5.  **Approximation to Normal Distribution for Academic Readiness:** The `Academic_Readiness_Score`'s distribution, when assessed against the 68-95-99.7 empirical rule, provides insights into its normality. If the actual percentages of scores within 1, 2, and 3 standard deviations closely align with the expected percentages, it indicates that `Academic_Readiness_Score` is approximately normally distributed. This has implications for using parametric statistical tests and for understanding the typical range of academic readiness in the student population.
